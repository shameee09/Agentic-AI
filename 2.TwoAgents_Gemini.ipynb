```python

import os
from dotenv import load_dotenv

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient

load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    raise ValueError("GEMINI_API_KEY not found in your .env file.")

code_writer_system_message = """
You have been given coding capability to solve tasks using Python code.

1. When you need to collect information, suggest complete Python or
shell code to retrieve and print the required information.
2. When you need to perform a task with code, provide complete,
executable code and print the relevant results.
3. Solve tasks step by step when necessary.
4. If no plan is provided, explain your plan first.
5. Clearly distinguish steps using code from steps using language skills.
6. Always indicate the script type in code blocks.
7. Do not provide incomplete code requiring the user to modify it.
8. Use print() when relevant.
9. Check the execution result returned by the user.
"""

model_client = OpenAIChatCompletionClient(
    model="gemini-3.5-flash",
    api_key=api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
    model_info={
        "vision": True,
        "function_calling": True,
        "json_output": True,
        "family": "unknown",
        "structured_output": True,
    },
)

# Agent 2: Code Writer
code_writer_agent = AssistantAgent(
    name="code_writer",
    model_client=model_client,
    system_message=code_writer_system_message,
)

print("Gemini Code Writer Agent created successfully!")
```

```python

from math import factorial
from collections import Counter

word = "ALGEBRA".lower()
counts = Counter(word)

result = factorial(len(word))

for count in counts.values():
    result //= factorial(count)

print(result)
```
