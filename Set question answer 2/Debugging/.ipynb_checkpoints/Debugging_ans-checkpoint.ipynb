```python
import os
from dotenv import load_dotenv

load_dotenv()

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")

if not GEMINI_API_KEY:
    raise ValueError("Gemini API key not found in .env file")

print("Gemini API key loaded successfully!")
```

```python
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(
    model="gemini-3.5-flash",
    api_key=GEMINI_API_KEY,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
    model_info={
        "vision": True,
        "function_calling": True,
        "json_output": True,
        "family": "unknown",
        "structured_output": True,
    }
)
```

```python
from autogen_agentchat.agents import AssistantAgent

toneAgent = AssistantAgent(
    name="ToneAgent",
    model_client=model_client,
    system_message=(
        "You are a professional communication assistant. "
        "Rewrite messages in a polite, professional, and friendly tone. "
        "Preserve the original meaning."
    )
)

message = "can u send me the file soon?"
result = await toneAgent.run(task=message)

tone_adjust = result.messages[-1].content

print("Original Message:", message)
print("Professional Message:", tone_adjust)
```

```python
from autogen_agentchat.agents import AssistantAgent

grammar_agent = AssistantAgent(
    name="GrammarFixer",
    model_client=model_client,
    system_message=(
        "You are a professional grammar correction assistant. "
        "Identify grammatical errors and correct the given sentence. "
        "Preserve the original meaning. "
        "Return only the corrected sentence."
    )
)

sentence = "he dont has no idea about the matter."

result = await grammar_agent.run(task=sentence)
corrected_sentence = result.messages[-1].content
print("Original Sentence:", sentence)
print("Corrected Sentence:", corrected_sentence)
```

```python
from autogen_agentchat.agents import AssistantAgent

summarizer = AssistantAgent(
    name="Summarizer",
    model_client=model_client,
    system_message=(
        "You are a customer review summarization assistant. "
        "Summarize customer reviews concisely while preserving "
        "the main issue and customer experience."
    )
)

sentimenter = AssistantAgent(
    name="SentimentAnalyzer",
    model_client=model_client,
    system_message=(
        "You are a sentiment analysis assistant. "
        "Analyze the given customer review summary and classify "
        "the sentiment as Positive, Negative, or Neutral. "
        "Briefly explain the reason."
    )
)

review_text = """
I ordered a laptop last week and it arrived with a broken screen.
Extremely disappointed!
"""

summary_result = await summarizer.run(task=review_text)
summary = summary_result.messages[-1].content

sentiment_result = await sentimenter.run(
    task=f"Analyze the sentiment of this customer review summary:\n{summary}"
)
sentiment = sentiment_result.messages[-1].content

print("Summary:", summary)
print("Sentiment:", sentiment)
```
